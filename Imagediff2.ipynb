```python
from PIL import Image, ImageChops
import matplotlib.pyplot as plt

# Load and convert images
#img1= Image.open("/content/sample_data/img/objs456_dist1_U_scene2.jpg").convert("RGB")
#img2 = Image.open("/content/sample_data/img/objs456_dist1_U_scene3.jpg").convert("RGB")

img1 = Image.open("/content/sample_data/img/IMG_9600.JPG").convert("RGB")
img2 = Image.open("/content/sample_data/img/IMG_9602.JPG").convert("RGB")


# Resize to match (if needed)
img2 = img2.resize(img1.size)

# Calculate the difference
diff = ImageChops.difference(img1, img2)
plt.figure(figsize=(18, 6))

plt.subplot(1, 3, 1)
plt.imshow(img1)
plt.title("Image 1")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(img2)
plt.title("Image 2")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(diff)
plt.title("Difference")
plt.axis("off")
```
